```python
df["bc"] = df["Links to Listing"].str.extract(r"bc=(\d+)")
print(df["bc"].duplicated().sum())

jnc_dupes = df[df["jnc"].duplicated(keep=False)]
non_exact = jnc_dupes[~jnc_dupes.duplicated(keep=False)]
print(len(non_exact))
non_exact.sort_values("jnc").head(20)
```

```python
from collections import Counter

cols = [c for c in df.columns if c not in ("jnc", "bc", "Links to Listing")]
counts = Counter()
for _, g in non_exact.groupby("jnc"):
    for c in cols:
        if g[c].nunique() > 1:
            counts[c] += 1
print(counts)
```

```python
phys = ["Floor", "Rooms", "Square Meters"]
conflict_ids = [
    jnc for jnc, g in non_exact.groupby("jnc")
    if any(g[c].nunique() > 1 for c in phys)
]
non_exact[non_exact["jnc"].isin(conflict_ids)].sort_values("jnc")
```
